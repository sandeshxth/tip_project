```python
from sklearn.preprocessing import MinMaxScaler
# Use select_dtypes to filter numerical columns
numerical_columns = data.select_dtypes(include=['int64', 'float64']).columns

# Count the number of numerical columns
num_numerical_columns = len(numerical_columns)

print("Number of numerical columns:", num_numerical_columns)
print("Numerical columns:", numerical_columns)

# Define numerical columns to normalize
numerical_columns = ['amt', 'city_pop', 'lat']

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Fit and transform the data
data_normalized = scaler.fit_transform(data[numerical_columns])

# Convert normalized data back to DataFrame
data_normalized = pd.DataFrame(data_normalized, columns=numerical_columns)

print("Normalized Data:")
print(data_normalized)
```

### Feature Engineering

```python
# Convert 'trans_date_trans_time' to datetime format
data['trans_date_trans_time'] = pd.to_datetime(data['trans_date_trans_time'])

# Transaction frequency per user
data['transaction_count'] = data.groupby('cc_num')['trans_date_trans_time'].transform('count')

# Average transaction amount per user
data['avg_transaction_amt'] = data.groupby('cc_num')['amt'].transform('mean')

# Time-based features
data['day_of_week'] = data['trans_date_trans_time'].dt.day_name()
data['hour_of_day'] = data['trans_date_trans_time'].dt.hour
data['month'] = data['trans_date_trans_time'].dt.month
data['season'] = (data['trans_date_trans_time'].dt.month % 12 + 3) // 3  # Calculate season based on month

# Print the updated dataset with new features
print(data.head())
```
